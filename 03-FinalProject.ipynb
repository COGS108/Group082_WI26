## Background and Prior Work

Airbnb has become a major platform in the short-term rental market and plays an important role in tourism and local housing economies. Nightly pricing is a key factor that affects both hosts and guests, as it influences host revenue, booking decisions, and perceived value for travelers. However, Airbnb prices are not set randomly; instead, they reflect a combination of listing characteristics such as location, property type, booking policies, and review activity.

Prior research has examined the determinants of Airbnb pricing using publicly available listing data. For example, Wang and Nicolau (2017) found that location, property type, and review ratings significantly influence nightly prices. Similarly, Gunter (2018) showed that listings with larger guest capacity, higher ratings, and more desirable property types tend to achieve higher prices.

In addition, Gibbs et al. (2018) applied a hedonic pricing model to Airbnb listings and demonstrated that a combination of listing characteristics — including location, property features, and demand-related signals — can explain a substantial portion of price variation. These studies suggest that regression-based approaches are effective for modeling Airbnb pricing and identifying key predictors.

While previous studies often examine large regional or cross-country markets, fewer focus specifically on pricing within a single metropolitan area such as San Diego. A city-level analysis allows us to better understand how local conditions and listing characteristics influence pricing in a specific urban context.

Building on this prior work, this project investigates how listing characteristics are associated with Airbnb nightly prices in San Diego. We focus on variables available in our dataset, including room type, neighborhood, geographic location (latitude and longitude), minimum nights, and review activity (number of reviews and last review). These variables capture key aspects of listing structure, location, and demand signals that may influence pricing.

We begin with exploratory data analysis (EDA) to examine the distribution of prices and relationships between variables, which informs the selection of predictors for subsequent regression modeling.

Because the dataset is observational, we interpret findings as associations rather than causal effects. Additionally, some factors that may influence pricing—such as interior design quality, host responsiveness, or detailed amenities—are not captured in the dataset. These limitations should be considered when interpreting the results.

#### Why We Use the 99th Percentile to Handle Outliers

To reduce the influence of extreme right-tail values, we removed listings priced above the 99th percentile (approximately $772 per night).

**Why 99th percentile instead of IQR?**
The interquartile range (IQR) method flags outliers based on the spread of the middle 50% of data. For a variable as skewed as Airbnb nightly price, IQR-based cutoffs can be overly aggressive — they may remove listings that are genuinely expensive but economically meaningful (e.g., large beachfront homes in La Jolla). The 99th percentile threshold is more conservative: it specifically targets the extreme upper tail while preserving the full range of typical listings, including high-end ones.

**Impact on data representativeness:**
This step removes fewer than 1% of observations. The resulting dataset still reflects the natural variation in the San Diego short-term rental market, from budget shared rooms to premium entire-home listings. However, it does mean our analyses and any future models may not generalize well to ultra-luxury listings above the threshold. We acknowledge this as a deliberate scope decision: our goal is to describe *typical* Airbnb pricing patterns, not extreme outliers.

### Relationship Between Listing Capacity and Nightly Price
To further examine the relationship between listing size and price, we fit a simple linear regression model with the number of guests a listing can accommodate as the predictor and the nightly price as the outcome. This allows us to quantify the strength of the association between listing capacity and Airbnb prices in San Diego.

````python
## YOUR CODE HERE
import statsmodels.api as sm

cols = [
    "nightly_price",
    "number_of_reviews"
]

df_model = listings[cols].dropna()
X = df_model[['number_of_reviews']]
y = df_model['nightly_price']

X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())
````

**Interpretation**

We fit a simple OLS regression model predicting nightly price from number of reviews. ## Ethics

### Data collection and representation bias

The Inside Airbnb dataset captures only listings that were *active and publicly visible* on Airbnb at the time of scraping. This creates a systematic **survivorship bias**: listings that have been deactivated, suspended, or voluntarily removed are not present in our data. Hosts who have exited the platform — potentially due to low demand, poor reviews, or local regulation — are invisible to our analysis. Our findings therefore describe the characteristics of *currently active* listings, not the full universe of San Diego short-term rentals.

Additionally, the geographic distribution of listings in our dataset is uneven. Coastal neighborhoods (La Jolla, Pacific Beach, Mission Beach, Coronado) are heavily represented, while inland and lower-income neighborhoods have comparatively fewer listings. This means our model's performance and interpretations may generalize less well to underrepresented areas of the city.

### Neighborhood as a proxy variable

One of our core predictors — neighborhood or ZIP code — is a geographic variable that, in the context of San Diego, carries substantial demographic and socioeconomic meaning. Neighborhoods in San Diego correlate with race and income due to historical patterns of residential segregation and redlining. When our model uses neighborhood as a predictor of price, it is not capturing "location" in a purely physical sense; it is also encoding information about the social composition of the area.

We do not claim that our model discriminates by race or income — we have no such variables in our dataset. But we acknowledge that neighborhood-based price predictions could, in practice, function as proxies for protected attributes. Any downstream use of our model (e.g., to guide pricing decisions) should account for this limitation explicitly.

### Review score bias

Our EDA includes review scores as a predictor variable. Review scores are generated only by guests who have booked and chosen to leave a review, which is a non-random subset of all guests. Guests who had very negative experiences may be less likely to write a review (preferring to simply not return), or conversely, more likely to write a negative one. This uncertainty means that review scores are not a neutral measure of quality — they reflect the opinions of a self-selected subset of guests, mediated by price expectations.

### Platform-level bias

Airbnb's search algorithm, pricing suggestions, and promotional features are not neutral. Listings that appear higher in search results may command higher prices not because of their intrinsic characteristics, but because of platform-mediated visibility advantages. Our dataset does not include information about listing visibility or algorithmic promotion, so we cannot account for this source of variation. Our findings should be understood as describing associations within the platform-visible listing pool, not as capturing the full causal structure of Airbnb pricing.

### Honest representation and scope of conclusions

Throughout our EDA and modeling, we commit to representing our findings honestly. Specifically:

- We do not make causal claims. All associations reported are correlational, derived from observational cross-sectional data.
- We acknowledge that our cleaned dataset excludes extreme outliers, which means our model may not generalize well to very high-end luxury listings or unusually cheap listings.
- We use median rather than mean prices where distributions are skewed, to avoid misleading readers about typical price levels.
- Our results describe pricing patterns on the Airbnb platform in San Diego as of the data collection date. They do not represent the full short-term or long-term rental market.
